## 3. ***Data Wrangling***

### Data Wrangling Code

````python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/Global Terrorism Data.csv", encoding='latin1')
df.rename(columns={
    'iyear': 'Year',
    'imonth': 'Month',
    'iday': 'Day',
    'country_txt': 'Country',
    'region_txt': 'Region',
    'city': 'City',
    'attacktype1_txt': 'AttackType',
    'target1': 'Target',
    'gname': 'Group',
    'weaptype1_txt': 'WeaponType',
    'nkill': 'Killed',
    'nwound': 'Wounded'
}, inplace=True)
# Fill numerical columns
df['Killed'] = df['Killed'].fillna(0)
df['Wounded'] = df['Wounded'].fillna(0)

# Fill categorical columns
df['Country'] = df['Country'].fillna("Unknown")
df['City'] = df['City'].fillna("Unknown")
df['AttackType'] = df['AttackType'].fillna("Unknown")
df['WeaponType'] = df['WeaponType'].fillna("Unknown")
df = df.drop_duplicates()
df['Casualties'] = df['Killed'] + df['Wounded']
# Replace unknown dates (0) with NaN
df['Month'] = df['Month'].replace(0, np.nan)
df['Day'] = df['Day'].replace(0, np.nan)
# Drop columns with too many missing values or not needed
df.drop(columns=['approxdate', 'resolution'], errors='ignore', inplace=True)
df['Year'] = df['Year'].astype(int)
df.info()
df.isnull().sum()
````

#### Chart - 14 - Correlation Heatmap

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,6)
df = pd.read_csv("/content/Global Terrorism Data.csv", encoding='latin1')
df.head()
df.rename(columns={'iyear': 'Year',
    'imonth': 'Month',
    'iday': 'Day',
    'country_txt': 'Country',
    'region_txt': 'Region',
    'attacktype1_txt': 'AttackType',
    'target1': 'Target',
    'weaptype1_txt': 'WeaponType',
    'nkill': 'Killed',
    'nwound': 'Wounded'
}, inplace=True)
df['Killed'] = df['Killed'].fillna(0)
df['Wounded'] = df['Wounded'].fillna(0)
df['Casualties'] = df['Killed'] + df['Wounded']
corr = df[['Killed', 'Wounded', 'Casualties']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
````

##### 1. Why did you pick the specific chart?

The correlation heatmap was selected because it provides a quick, clear, and visually intuitive understanding of relationships between variables, making it highly useful for analysis and decision-making.

##### 2. What is/are the insight(s) found from the chart?

The heatmap reveals strong relationships among impact-related variables, helping in understanding how different factors contribute to the severity of terrorist incidents and guiding better analytical decisions.
